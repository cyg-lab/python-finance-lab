```python
import pandas as pd

transactions = [
    {"account": "매출", "amount": 1200000},
    {"account": "광고비", "amount": 800000},
    {"account": "접대비", "amount": 450000},
    {"account": "매출", "amount": 1500000}
]

df = pd.DataFrame(transactions)

filtered = df[df["amount"] >= 700000]

result = (
    filtered
    .groupby("account")["amount"]
    .agg(["sum", "count", "mean"])
    .reset_index()
    .sort_values("sum", ascending=False)
)

print(result)

result.to_csv("result.csv", index=False)
```

```python
df["date"] = pd.to_datetime(df["date"])
df["month"] = df["date"].dt.month
df.groupby("month")["amount"].sum()
```

```python
result = (
    df.groupby(["month", "account"])["amount"]
    .sum()
    .reset_index()
)

print(result)
```

```python
summary = (
    df.groupby("account")["amount"]
    .agg(["sum", "count", "mean"])
    .reset_index()
    .sort_values("mean", ascending=False)
)
summary["mean"] = summary["mean"].round(0).astype(int)
print(summary)
```

```python
high_entertainment = df[
    (df["amount"] >= 500000)
    & (df["account"] == "접대비")
]

print(high_entertainment)
```

```python
df[df["account"] == "접대비"]["amount"].mean()
average = df[df["account"] == "접대비"]["amount"].mean()

print(round(average))
```

```python
filtered = df[(df['account']== '접대비')&(df['amount']>=average)]

print(filtered)
```

```python
df["account_mean"] = df.groupby("account")["amount"].transform("mean")
df["high_amount"] = df["amount"] >= 1_000_000
df["above_mean"] = df["amount"] > df["account_mean"]

df["review_needed"] = df["high_amount"] & df["above_mean"]

review_df = df[df["review_needed"]]

print(review_df)

```

```python
df["amount_ratio"] = (
    df["amount"] / df["account_mean"]
).round(2)
result = df[df["amount_ratio"] >= 1.5]

print(result)
```

```python
df["review_needed"] = (
    (df["amount"] >= 1_000_000)
    | (df["amount_ratio"] >= 1.5)
)

print(df["review_needed"])
```

```python
review_df = df[df["review_needed"]]

print(review_df[["date", "account", "vendor", "amount", "amount_ratio"]])
```

```python
expense_review = df[
    df["review_needed"]
    & (df["account"] != "매출")
]

print(expense_review[["date", "account", "vendor", "amount", "amount_ratio"]])
```
